```python
import pandas as pd
import openpyxl
```

```python
path = "CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"
```

```python
populacao_bruto = pd.read_excel(path, header=None)
```

```python
populacao_bruto
```

```python
populacao_total = populacao_bruto.iloc[3:5573, 1:].copy()
```

```python
populacao_estado = populacao_total
```

```python
populacao_total.columns = ['UF', 'cod_uf', 'cod_munic', 'municipio', 'pop_2010_sinopse', 'pop_2010', 'pop_2022']

populacao_total['pop_2010'] = pd.to_numeric(populacao_total['pop_2010'])
populacao_total['pop_2022'] = pd.to_numeric(populacao_total['pop_2022'])

populacao_total
```

```python
populacao_estado = populacao_total.groupby('UF')[['pop_2010', 'pop_2022']].sum().reset_index()

populacao_estado['crescimento'] = populacao_estado['pop_2022'] - populacao_estado['pop_2010']
populacao_estado = populacao_estado.sort_values(by='crescimento', ascending=False)

print(populacao_estado)

populacao_estado.to_csv('populacao_estado_crescimento.csv', index=False, sep=';', encoding='utf-8-sig')
```

```python
populacao_municipio = populacao_total[['UF', 'municipio', 'pop_2010', 'pop_2022']].copy()

populacao_municipio['crescimento'] = populacao_municipio['pop_2022'] - populacao_municipio['pop_2010']
populacao_municipio = populacao_municipio.sort_values(by='crescimento', ascending=False)

print(populacao_municipio.head(15))

populacao_municipio.to_csv('populacao_municipio_crescimento.csv', index=False, sep=';', encoding='utf-8-sig')
```
